##### Reverse a string

```python
s = "Ritik"
print(s[::-1])
def rev(s):
    rev = ""
    for i in s:
        rev = i + rev
    return(rev)
print(rev(s))
```

##### Check palindrome string

```python
s = "abba"
def palindrome(s):
    if s == s[::-1]:
        return True
    else:
        False
print(palindrome(s))
```
