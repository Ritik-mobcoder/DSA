## Q1. Two Sum

**Difficulty:** Easy

**Problem:**  
Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.

**Example:**
```
Input:  nums = [2, 7, 11, 15], target = 9
Output: [0, 1]
```

**Concepts:** Hashing

````python
# Write your solution here
def two_sum(lst, k):
    hashmap = {}
    for i in range(len(lst)):
        comp = k - lst[i]
        if comp in hashmap:
            return [hashmap[comp], i]
        hashmap[lst[i]] = i

lst = [2, 7, 11, 15]
target = 9
print(two_sum(lst, target))
````

## Q2. Best Time to Buy and Sell Stock

**Difficulty:** Easy

**Problem:**  
Given an array `prices` where `prices[i]` is the price of a stock on day `i`, return the maximum profit you can achieve. If no profit is possible, return `0`.

**Example:**
```
Input:  prices = [7, 1, 5, 3, 6, 4]
Output: 5
```

**Concepts:** Sliding Window, Greedy

````python
# Write your solution here
def maxProfit(lst):
        min_price = lst[0]
        max_profit = 0

        for price in lst:
            if price < min_price:
                min_price = price

            profit = price - min_price

            if profit > max_profit:
                max_profit = profit

        return max_profit
prices = [7, 1, 5, 3, 6, 4]
print(maxProfit(prices))
````

## Q3. Maximum Subarray (Kadane's Algorithm)

**Difficulty:** Medium

**Problem:**  
Given an integer array `nums`, find the contiguous subarray with the largest sum and return its sum.

**Example:**
```
Input:  nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
Output: 6  (subarray: [4, -1, 2, 1])
```

**Concepts:** Kadane's Algorithm, Dynamic Programming

````python
# Write your solution here

def maxSubArray(nums):
    current_sum = max_sum = nums[0]
    start = end = temp_start = 0

    for i in range(1, len(nums)):
        if nums[i] > current_sum + nums[i]:
            current_sum = nums[i]
            temp_start = i
        else:
            current_sum += nums[i]

        if current_sum > max_sum:
            max_sum = current_sum
            start = temp_start
            end = i

    return max_sum, nums[start:end + 1]
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(maxSubArray(nums))
````

## Q4. Move Zeroes

**Difficulty:** Easy

**Problem:**  
Given an integer array `nums`, move all `0`s to the end while maintaining the relative order of the non-zero elements. Do it in-place.

**Example:**
```
Input:  nums = [0, 1, 0, 3, 12]
Output: [1, 3, 12, 0, 0]
```

**Concepts:** Two Pointers

````python
# Write your solution here
def move_zero(nums):
    i =0
    for j in range(len(nums)):
        if nums[j] != 0:
            nums[i], nums[j] = nums[j], nums[i]
            i+=1
    return nums
        

nums = [0, 1, 0, 3, 12]
print(move_zero(nums))
````
