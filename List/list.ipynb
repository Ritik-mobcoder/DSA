##### Find the maximum and minimum element in an array

```python
lst = [4,3,68,35,6,322,56,79]
```

```python
def max_min(lst):
    d ={"minimum" :lst[0], "maximum": lst[0]}

    for i in range(0, len(lst)):
        if lst[i] < d.get("minimum"):
            d["minimum"] = lst[i]
        if lst[i] > d.get("maximum"):
            d["maximum"] = lst[i]
    return d

        
print(max_min(lst))
```

##### Reverse an array

```python
def  reverse_lst(arr):
    left = 0
    right = len(arr) - 1
    while left < right:

        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr

# l = []

lst = [4,3,68,35,6,322,56,79]
print(reverse_lst(lst))
```

##### Find the second largest element

```python
def find_second_largest(lst):
    largest = float('-inf')
    second_largest = float('-inf')
    for i in lst:
        if i > largest:
            second_largest = largest
            largest = i
        
        elif i > second_largest and i != largest:
            second_largest = i
    return second_largest  if second_largest != float('-inf') else -1

lst = [79, 56, 322, 6, 35, 68, 3, 4]

print(find_second_largest(lst))


def second_largest(lst):
    largest = float('inf')
    second_largest = float('-inf')
    for i in lst:
        if i > largest:
            second_largest = largest
            largest = i
        elif i > second_largest and i!= largest:
            second_largest = i
    return second_largest if second_largest != float('-inf') else -1
```

##### Check if array is sorted

```python
def is_sorted(lst):
    if len(lst)<=1:
        return True
    for i in range(len(lst)-1):
        if lst[i]> lst[i+1]:
            return False
    return True
print(is_sorted(lst))
```

##### Remove duplicates from sorted array

```python
def removeDuplicates(lst):
    if not lst:
        return 0, []

    i = 0

    for j in range(1, len(lst)):
        if lst[j] != lst[i]:
            i += 1
            lst[i] = lst[j]

    return i + 1, lst[:i + 1]
lst = [1,1,2,2,3,4,4]
print(removeDuplicates(lst))
```

##### Rotate array left/right by K steps

```python
def rotateLst(lst, k, rotate= "right"):
    n = len(lst)
    k = k % n
    if rotate == "right":
        return lst[-k:] + lst[:-k]
    elif rotate == "left":
       return lst[k:] + lst[:k]
lst = [1,0,34,3,0,4,2,0,2]
print(rotateLst(lst, 2, "left"))
```

##### Two Sum problem

```python
def twoSum(lst, k):
    for i in range(0,len(lst)):
        for j in range(len(lst)):
            if lst[i] + lst[j] ==k:
                return i,j 
lst = [34, 3, 0, 4, 2, 0, 2, 1, 0]
print(twoSum(lst, 4))

# optimal solution
def twoSumBest(nums, k):
    hashmap = {}
    for i in range(len(nums)):
        complement = k - nums[i]

        if complement in hashmap:
            return [hashmap[complement], i]

        hashmap[nums[i]] = i
print(twoSumBest(lst,4))
```
