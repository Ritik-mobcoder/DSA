##### Find the maximum and minimum element in an array

```python
lst = [4,3,68,35,6,322,56,79]
```

```python
def max_min(lst):
    d ={"minimum" :lst[0], "maximum": lst[0]}

    for i in range(0, len(lst)):
        if lst[i] < d.get("minimum"):
            d["minimum"] = lst[i]
        if lst[i] > d.get("maximum"):
            d["maximum"] = lst[i]
    return d

print(max_min(lst))
```

##### Check if array is sorted

```python
def is_sorted(lst):
    if len(lst)<=1:
        return True
    for i in range(len(lst)-1):
        if lst[i]> lst[i+1]:
            return False
    return True
print(is_sorted(lst))
```

##### Remove duplicates from sorted array

```python
def removeDuplicates(nums):
    if not nums:
        return 0, []

    i = 0

    for j in range(1, len(nums)):
        if nums[j] != nums[i]:
            i += 1
            nums[i] = nums[j]

    return i + 1, nums[:i + 1]
lst = [1,1,2,2,3,4,4]
print(removeDuplicates(lst))
```

##### Move all zeros to the end

```python
def moveZeroes(lst):
    i = 0
    for j in range(0, len(lst)):
        if lst[j] != 0:
            lst[i], lst[j] = lst[j], lst[i]
            i+=1
    return lst

lst = [1,0,34,3,0,4,2,0,2]
print(moveZeroes(lst))
```

##### Rotate array left/right by K steps

```python
def rotateLst(lst, k, rotate= "right"):
    n = len(lst)
    k = k % n
    if rotate == "right":
        return lst[-k:] + lst[:-k]
    elif rotate == "left":
       return lst[k:] + lst[:k]
lst = [1,0,34,3,0,4,2,0,2]
print(rotateLst(lst, 2, "left"))
```

##### Two Sum problem

```python
def twoSum(lst, k):
    for i in range(0,len(lst)):
        for j in range(len(lst)):
            if lst[i] + lst[j] ==k:
                return i,j 
lst = [34, 3, 0, 4, 2, 0, 2, 1, 0]
print(twoSum(lst, 4))

# optimal solution
def twoSumBest(nums, k):
    hashmap = {}
    for i in range(len(nums)):
        complement = k - nums[i]

        if complement in hashmap:
            return [hashmap[complement], i]

        hashmap[nums[i]] = i
print(twoSumBest(lst,4))
```
